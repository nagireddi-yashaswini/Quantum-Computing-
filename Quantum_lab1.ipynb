# 4. Introduction

Quantum computing is a computing paradigm based on the principles of quantum mechanics. Classical computers process information using bits that can represent either 0 or 1. Quantum computers use quantum bits, called qubits, which can exist in a combination of computational basis states.

A general single-qubit state is represented as:

\[
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle
\]

where \(\alpha\) and \(\beta\) are complex probability amplitudes satisfying:

\[
|\alpha|^2 + |\beta|^2 = 1
\]

Quantum circuits use quantum gates to manipulate qubits. Gates such as Pauli-X, Pauli-Y, Pauli-Z and Hadamard operate on individual qubits, while gates such as CNOT, CZ, SWAP and Toffoli operate on multiple qubits.

Quantum computing also supports important phenomena such as superposition and entanglement. Bell and GHZ states are examples of entangled quantum states.

Measurements convert quantum information into classical outcomes. Because quantum measurement is probabilistic, repeated measurements are used to estimate the probability distribution of possible outcomes.

In this project, Qiskit is used to create, visualize, and simulate fundamental quantum circuits.

# 5. Mathematical Background

## 5.1 Computational Basis States

The two computational basis states of a qubit are:

\[
|0\rangle =
\begin{bmatrix}
1\\
0
\end{bmatrix}
\]

and

\[
|1\rangle =
\begin{bmatrix}
0\\
1
\end{bmatrix}
\]

## 5.2 Superposition

A qubit can exist in a superposition:

\[
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle
\]

with:

\[
|\alpha|^2 + |\beta|^2 = 1
\]

## 5.3 Pauli Gates

The Pauli-X gate is:

\[
X =
\begin{bmatrix}
0&1\\
1&0
\end{bmatrix}
\]

The Pauli-Y gate is:

\[
Y =
\begin{bmatrix}
0&-i\\
i&0
\end{bmatrix}
\]

The Pauli-Z gate is:

\[
Z =
\begin{bmatrix}
1&0\\
0&-1
\end{bmatrix}
\]

## 5.4 Hadamard Gate

The Hadamard gate is:

\[
H =
\frac{1}{\sqrt{2}}
\begin{bmatrix}
1&1\\
1&-1
\end{bmatrix}
\]

It creates an equal superposition from a computational basis state.

## 5.5 Measurement Probability

If:

\[
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle
\]

then:

\[
P(0)=|\alpha|^2
\]

and

\[
P(1)=|\beta|^2
\]

## 5.6 Entanglement

For two qubits, the Bell state used in this project is:

\[
|\Phi^+\rangle =
\frac{|00\rangle+|11\rangle}{\sqrt{2}}
\]

The GHZ state is:

\[
|GHZ\rangle =
\frac{|000\rangle+|111\rangle}{\sqrt{2}}
\]

# 6. Circuit Design Methodology

The quantum circuits in this project are designed and simulated using the following procedure:

1. Initialize the required number of qubits.
2. Apply the required quantum gates.
3. Obtain the exact quantum state using a statevector simulation.
4. Calculate the theoretical probability distribution.
5. Add measurements to a copy of the circuit.
6. Execute repeated measurements using a specified number of shots.
7. Record the measurement counts.
8. Visualize the circuit and quantum state.
9. Compare theoretical probabilities with simulated results.
10. Analyze the circuit using Qiskit transpilation.

The original circuit is kept without measurement so that the quantum state can be analyzed before measurement. A separate copy is used for measurement simulation.

# 7. Qiskit Implementation

Qiskit is used as the primary quantum computing framework. The `QuantumCircuit` class is used to construct quantum circuits, while `Statevector` is used to obtain the exact quantum state before measurement.

Qiskit's visualization tools are used for Bloch-sphere, Q-sphere, and histogram visualizations.

The circuits are simulated using ideal statevector sampling with a configurable number of shots.

````python
from math import pi
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import (
    plot_bloch_multivector, plot_state_qsphere, plot_histogram
)

def simulate_ideal(circuit, shots=1024, seed=7):
    # Inspect the pure state before measurement.
    state = Statevector.from_instruction(circuit)

    # Sample a measured copy; keep the original unitary circuit intact.
    measured = circuit.copy()
    measured.measure_all()
    pub_result = (
        StatevectorSampler(seed=seed)
        .run([measured], shots=shots)
        .result()[0]
    )
    counts = pub_result.data.meas.get_counts()
    return state, counts
````

# 7.2 Multi-Qubit Circuits

This section implements multi-qubit quantum circuits including CNOT, CZ, SWAP, Toffoli, Bell-state, and GHZ-state circuits.

## 7.2.1 Controlled-NOT (CNOT / CX)

The CNOT gate uses one qubit as a control and another as a target.

If the control qubit is |1⟩, the target qubit is flipped.

````python
qc = QuantumCircuit(2)
qc.x(0)          # q0 becomes control value 1
qc.cx(0, 1)      # control q0, target q1
display(qc.draw("mpl"))
state, counts = simulate_ideal(qc)
print(state.probabilities_dict())
print(counts)
````

The Controlled-Z gate applies a phase flip to the |11⟩ component of a two-qubit state.

It is a two-qubit controlled phase operation.

````python
qc = QuantumCircuit(2)
qc.h([0, 1])
qc.cz(0, 1)
display(qc.draw("mpl"))
state, counts = simulate_ideal(qc)
print(state.data)
print(counts)
````

The SWAP gate exchanges the quantum states of two qubits.

## 7.2.4 Toffoli / Controlled-Controlled-NOT Circuit

````python
qc = QuantumCircuit(3)
qc.x([0, 1])
qc.ccx(0, 1, 2)
display(qc.draw("mpl"))
state, counts = simulate_ideal(qc)
print(state.probabilities_dict())
print(counts)
````

# 11. Transpilation Analysis

Transpilation converts a quantum circuit into a form suitable for execution on a target quantum backend.

The transpilation process can optimize the circuit and change its gate representation and circuit depth.

In this project, the original circuit and transpiled circuit are compared using:

- Circuit depth
- Gate count
- Gate types
- Transpiled circuit diagram

````python
from qiskit import transpile

# Bell circuit
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)

transpiled_bell = transpile(bell, optimization_level=3)

print("Original circuit:")
display(bell.draw("mpl"))

print("Transpiled circuit:")
display(transpiled_bell.draw("mpl"))

print("Original depth:", bell.depth())
print("Transpiled depth:", transpiled_bell.depth())

print("Original gates:", bell.count_ops())
print("Transpiled gates:", transpiled_bell.count_ops())
````

# 12. Functional Circuit Designer Interface

A Streamlit-based graphical interface is provided as part of the project. The interface allows users to select a circuit, specify rotation parameters and measurement shots, visualize the circuit, inspect the statevector, view probabilities, and analyze measurement results.

qiskit[all]~=2.5.1

````python
from math import pi
import streamlit as st
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import (
    plot_bloch_multivector,
    plot_state_qsphere,
    plot_histogram,
)
CIRCUITS = [
    "Identity (I)", "Pauli-X", "Pauli-Y", "Pauli-Z",
    "Hadamard (H)", "Phase-S", "Phase-T",
    "Rx", "Ry", "Rz", "Measurement demo",
    "CNOT / CX", "Controlled-Z", "SWAP",
    "Toffoli / CCX", "Bell state", "GHZ state",
]

def build_circuit(name: str, theta: float) -> QuantumCircuit:
    if name in {
        "Identity (I)", "Pauli-X", "Pauli-Y", "Pauli-Z",
        "Hadamard (H)", "Phase-S", "Phase-T",
        "Rx", "Ry", "Rz", "Measurement demo",
    }:
        qc = QuantumCircuit(1, name=name)
    elif name in {"CNOT / CX", "Controlled-Z", "SWAP", "Bell state"}:
        qc = QuantumCircuit(2, name=name)
    else:
        qc = QuantumCircuit(3, name=name)

    if name == "Identity (I)":
        qc.id(0)
    elif name == "Pauli-X":
        qc.x(0)
    elif name == "Pauli-Y":
        qc.y(0)
    elif name == "Pauli-Z":
        qc.h(0); qc.z(0)       # exposes phase on |+>
    elif name == "Hadamard (H)":
        qc.h(0)
    elif name == "Phase-S":
        qc.h(0); qc.s(0)
    elif name == "Phase-T":
        qc.h(0); qc.t(0)
    elif name == "Rx":
        qc.rx(theta, 0)
    elif name == "Ry":
        qc.ry(theta, 0)
    elif name == "Rz":
        qc.h(0); qc.rz(theta, 0)
    elif name == "Measurement demo":
        qc.h(0)                # measurement is added later
    elif name == "CNOT / CX":
        qc.x(0); qc.cx(0, 1)
    elif name == "Controlled-Z":
        qc.h([0, 1]); qc.cz(0, 1)
    elif name == "SWAP":
        qc.x(0); qc.swap(0, 1)
    elif name == "Toffoli / CCX":
        qc.x([0, 1]); qc.ccx(0, 1, 2)
    elif name == "Bell state":
        qc.h(0); qc.cx(0, 1)
    elif name == "GHZ state":
        qc.h(0); qc.cx(0, 1); qc.cx(0, 2)
    return qc
````

````python
def simulate(circuit: QuantumCircuit, shots: int, seed: int = 7):
    # Exact pre-measurement state.
    state = Statevector.from_instruction(circuit)

    # Add terminal measurements to a separate copy.
    measured = circuit.copy()
    measured.measure_all()

    # V2 sampler returns a PubResult with a BitArray named 'meas'.
    pub_result = (
        StatevectorSampler(seed=seed)
        .run([measured], shots=shots)
        .result()[0]
    )
    counts = pub_result.data.meas.get_counts()
    return state, measured, counts
````

````python
st.set_page_config(
    page_title="Quantum Circuit Designer",
    page_icon="⚛",
    layout="wide",
)

st.title("Quantum Circuit Designer and Simulator")
st.caption("Build → inspect state → measure → interpret")

with st.sidebar:
    name = st.selectbox("Circuit", CIRCUITS)
    theta = st.slider(
        "Rotation angle θ (radians)",
        min_value=0.0, max_value=2*pi, value=pi/2, step=pi/16,
    )
    shots = st.select_slider(
        "Shots", options=[32, 128, 512, 1024, 4096, 8192], value=1024
    )

circuit = build_circuit(name, theta)
state, measured, counts = simulate(circuit, shots)

left, right = st.columns(2)
with left:
    st.subheader("Unitary circuit")
    st.pyplot(circuit.draw("mpl"))
    st.subheader("Measured circuit")
    st.pyplot(measured.draw("mpl"))

with right:
    st.subheader("State before measurement")
    st.code(str(state.data), language="text")
    st.write("Ideal probabilities:", state.probabilities_dict())
    st.subheader("Sampled counts")
    st.write(counts)
    st.pyplot(plot_histogram(counts))

st.subheader("State visualization")
if circuit.num_qubits == 1:
    st.pyplot(plot_bloch_multivector(state))
else:
    st.pyplot(plot_state_qsphere(state))

st.info(
    "Counts fluctuate because they are sampled. "
    "Statevector probabilities are the exact ideal prediction."
)
````

# 13. Challenges and Limitations

The main challenges and limitations of the project include:

- Quantum measurement is probabilistic, so finite-shot simulations may differ from theoretical probabilities.
- Statevector simulation becomes computationally expensive as the number of qubits increases.
- The current application focuses on ideal quantum simulation.
- Real quantum hardware introduces noise, decoherence, gate errors, and measurement errors.
- The circuit designer currently focuses on predefined fundamental circuits rather than arbitrary circuit construction.
- Real quantum hardware access may be limited by queue time and backend availability.

````python
from math import isclose

def probabilities(circuit):
    return Statevector.from_instruction(circuit).probabilities_dict()

# Deterministic X test
x = QuantumCircuit(1); x.x(0)
assert isclose(float(probabilities(x).get('1', 0)), 1.0)

# Bell support test
bell = QuantumCircuit(2); bell.h(0); bell.cx(0, 1)
p = probabilities(bell)
assert set(p) == {'00', '11'}
assert isclose(float(p['00']), 0.5, abs_tol=1e-12)
assert isclose(float(p['11']), 0.5, abs_tol=1e-12)

# GHZ support test
ghz = QuantumCircuit(3); ghz.h(0); ghz.cx(0,1); ghz.cx(0,2)
p = probabilities(ghz)
assert set(p) == {'000', '111'}
````

````python
from math import pi
import streamlit as st
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import (
    plot_bloch_multivector, plot_state_qsphere, plot_histogram
)

def simulate_ideal(circuit, shots=1024, seed=7):
    # Inspect the pure state before measurement.
    state = Statevector.from_instruction(circuit)

    # Sample a measured copy; keep the original unitary circuit intact.
    measured = circuit.copy()
    measured.measure_all()
    pub_result = (
        StatevectorSampler(seed=seed)
        .run([measured], shots=shots)
        .result()[0]
    )
    counts = pub_result.data.meas.get_counts()
    return state, counts

CIRCUITS = [
    "Identity (I)", "Pauli-X", "Pauli-Y", "Pauli-Z",
    "Hadamard (H)", "Phase-S", "Phase-T",
    "Rx", "Ry", "Rz", "Measurement demo",
    "CNOT / CX", "Controlled-Z", "SWAP",
    "Toffoli / CCX", "Bell state", "GHZ state",
]

def build_circuit(name: str, theta: float) -> QuantumCircuit:
    if name in {
        "Identity (I)", "Pauli-X", "Pauli-Y", "Pauli-Z",
        "Hadamard (H)", "Phase-S", "Phase-T",
        "Rx", "Ry", "Rz", "Measurement demo",
    }:
        qc = QuantumCircuit(1, name=name)
    elif name in {"CNOT / CX", "Controlled-Z", "SWAP", "Bell state"}:
        qc = QuantumCircuit(2, name=name)
    else:
        qc = QuantumCircuit(3, name=name)

    if name == "Identity (I)":
        qc.id(0)
    elif name == "Pauli-X":
        qc.x(0)
    elif name == "Pauli-Y":
        qc.y(0)
    elif name == "Pauli-Z":
        qc.h(0); qc.z(0)       # exposes phase on |+>
    elif name == "Hadamard (H)":
        qc.h(0)
    elif name == "Phase-S":
        qc.h(0); qc.s(0)
    elif name == "Phase-T":
        qc.h(0); qc.t(0)
    elif name == "Rx":
        qc.rx(theta, 0)
    elif name == "Ry":
        qc.ry(theta, 0)
    elif name == "Rz":
        qc.h(0); qc.rz(theta, 0)
    elif name == "Measurement demo":
        qc.h(0)                # measurement is added later
    elif name == "CNOT / CX":
        qc.x(0); qc.cx(0, 1)
    elif name == "Controlled-Z":
        qc.h([0, 1]); qc.cz(0, 1)
    elif name == "SWAP":
        qc.x(0); qc.swap(0, 1)
    elif name == "Toffoli / CCX":
        qc.x([0, 1]); qc.ccx(0, 1, 2)
    elif name == "Bell state":
        qc.h(0); qc.cx(0, 1)
    elif name == "GHZ state":
        qc.h(0); qc.cx(0, 1); qc.cx(0, 2)
    return qc

def simulate(circuit: QuantumCircuit, shots: int, seed: int = 7):
    # Exact pre-measurement state.
    state = Statevector.from_instruction(circuit)

    # Add terminal measurements to a separate copy.
    measured = circuit.copy()
    measured.measure_all()

    # V2 sampler returns a PubResult with a BitArray named 'meas'.
    pub_result = (
        StatevectorSampler(seed=seed)
        .run([measured], shots=shots)
        .result()[0]
    )
    counts = pub_result.data.meas.get_counts()
    return state, measured, counts

def probabilities(circuit):
    return Statevector.from_instruction(circuit).probabilities_dict()

# Streamlit app logic (removed st.set_page_config as it cannot be run here)
# These parts are typically run in a separate .py file for Streamlit apps.
# For the purpose of consolidation, they are included, but marked as app logic.
# To run the Streamlit app, save this combined code into a file (e.g., app.py)
# and execute 'streamlit run app.py' in the terminal.

# Uncomment the following block and save to app.py to run the Streamlit app
# if __name__ == '__main__':
#     st.set_page_config(
#         page_title="Quantum Circuit Designer",
#         page_icon="⚛",
#         layout="wide",
#     )

#     st.title("Quantum Circuit Designer and Simulator")
#     st.caption("Build → inspect state → measure → interpret")

#     with st.sidebar:
#         name = st.selectbox("Circuit", CIRCUITS)
#         theta = st.slider(
#             "Rotation angle θ (radians)",
#             min_value=0.0, max_value=2*pi, value=pi/2, step=pi/16,
#         )
#         shots = st.select_slider(
#             "Shots", options=[32, 128, 512, 1024, 4096, 8192], value=1024
#         )

#     circuit = build_circuit(name, theta)
#     state, measured, counts = simulate(circuit, shots)

#     left, right = st.columns(2)
#     with left:
#         st.subheader("Unitary circuit")
#         st.pyplot(circuit.draw("mpl"))
#         st.subheader("Measured circuit")
#         st.pyplot(measured.draw("mpl"))

#     with right:
#         st.subheader("State before measurement")
#         st.code(str(state.data), language="text")
#         st.write("Ideal probabilities:", state.probabilities_dict())
#         st.subheader("Sampled counts")
#         st.write(counts)
#         st.pyplot(plot_histogram(counts))

#     st.subheader("State visualization")
#     if circuit.num_qubits == 1:
#         st.pyplot(plot_bloch_multivector(state))
#     else:
#         st.pyplot(plot_state_qsphere(state))

#     st.info(
#         "Counts fluctuate because they are sampled. "
#         "Statevector probabilities are the exact ideal prediction."
#     )

print("Source code consolidated successfully. The Streamlit app logic is commented out and should be run in a separate file if needed.")
````

````python
from qiskit import transpile
from qiskit_aer import AerSimulator
from qiskit import QuantumCircuit

# Define a quantum circuit to be used for transpilation (e.g., Bell state)
bell_circuit = QuantumCircuit(2, name="Bell State Circuit")
bell_circuit.h(0)
bell_circuit.cx(0, 1)

print("Original Bell State Circuit defined:")
print(bell_circuit.draw('text'))

# Define a generic backend for transpilation
simulator = AerSimulator()

# Transpile the circuit
transpiled_circuit = transpile(bell_circuit, simulator)

print("\nTranspiled Bell State Circuit defined:")
print(transpiled_circuit.draw('text'))

# Analyze the transpiled circuit
print(f"\nOriginal circuit depth: {bell_circuit.depth()}")
print(f"Transpiled circuit depth: {transpiled_circuit.depth()}")

print(f"\nOriginal circuit gate count: {bell_circuit.size()}")
print(f"Transpiled circuit gate count: {transpiled_circuit.size()}")

print(f"\nOriginal circuit operations: {bell_circuit.count_ops()}")
print(f"Transpiled circuit operations: {transpiled_circuit.count_ops()}")
````

### Transpilation Analysis Results

The transpilation analysis for the Bell State Circuit shows that both the original and transpiled circuits are identical in terms of circuit depth, gate count, and the types of operations. This is an expected outcome for a simple circuit like the Bell state when transpiling for a generic simulator backend such as `AerSimulator`, as there might not be significant opportunities for optimization or changes in the gate set.

In more complex quantum circuits or when targeting specific hardware backends with different native gate sets and connectivity constraints, the transpilation process would typically lead to noticeable changes in the circuit structure, such as increased depth, different gate counts, or the insertion of SWAP gates to satisfy qubit connectivity requirements.
